## 3. Tratamento dos dados

### 3.1 Removendo espaços em branco nas colunas de texto

```python
for col in df.select_dtypes(include=['object']).columns:
    df[col] = df[col].str.strip()
```

### 3.2 Traduzindo para o português

```python
trad_pais = {
    'China': 'China', 'India': 'Índia', 'UK': 'Reino Unido',
    'Germany': 'Alemanha', 'France': 'França', 'Australia': 'Austrália',
    'Russia': 'Rússia', 'Brazil': 'Brasil', 'Japan': 'Japão', 'USA': 'EUA'
}
```

```python
trad_ataque = {
    'Phishing': 'Phishing', 'Ransomware': 'Ransomware',
    'Man-in-the-Middle': 'Homem no Meio (MitM)',
    'DDoS': 'Negação de Serviço (DDoS)', 'SQL Injection': 'Injeção SQL',
    'Malware': 'Malware'
}
```

```python
trad_industria = {
    'Education': 'Educação', 'Retail': 'Varejo', 'IT': 'TI / Tecnologia',
    'Telecommunications': 'Telecomunicações', 'Government': 'Governo',
    'Banking': 'Bancário / Financeiro', 'Healthcare': 'Saúde'
}
```

```python
trad_origem = {
    'Hacker Group': 'Grupo de Hackers', 'Nation-state': 'Estado-Nação',
    'Insider': 'Colaborador Interno', 'Unknown': 'Desconhecido'
}
```

```python
trad_vulnerabilidade = {
    'Unpatched Software': 'Software Desatualizado',
    'Weak Passwords': 'Senhas Fracas',
    'Social Engineering': 'Engenharia Social',
    'Zero-day': 'Falha Dia Zero (Zero-Day)'
}
```

```python
trad_defesa = {
    'VPN': 'VPN', 'Firewall': 'Firewall',
    'AI-based Detection': 'Detecção por IA',
    'Antivirus': 'Antivírus', 'Encryption': 'Criptografia'
}
```

```python
df['Country'] = df['Country'].replace(trad_pais)
df['Attack Type'] = df['Attack Type'].replace(trad_ataque)
df['Target Industry'] = df['Target Industry'].replace(trad_industria)
df['Attack Source'] = df['Attack Source'].replace(trad_origem)
df['Security Vulnerability Type'] = df['Security Vulnerability Type'].replace(trad_vulnerabilidade)
df['Defense Mechanism Used'] = df['Defense Mechanism Used'].replace(trad_defesa)
```

```python
novos_nomes = {
    'Country': 'País',
    'Year': 'Ano',
    'Attack Type': 'Tipo de Ataque',
    'Target Industry': 'Setor Alvo',
    'Financial Loss (in Million $)' : 'Prejuízo Financeiro (Milhões $)',
    'Number of Affected Users': 'Usuários Afetados',
    'Attack Source': 'Origem do Ataque',
    'Security Vulnerability Type': 'Vulnerabilidade',
    'Defense Mechanism Used': 'Mecanismo de Defesa',
    'Incident Resolution Time (in Hours)': 'Tempo de Resolução (Horas)'
}
```

```python
df = df.rename(columns=novos_nomes)
```

```python
print("--- Primeiras linhas dos dados ---")
display(df.head())
```

### 3.3 Criação de Indicadores para Tomada de Decisão (Engenharia de Atributos)

```python
# 1. Criando a coluna 'Nível de Impacto' baseada em quartis financeiros
limite_alto = df['Prejuízo Financeiro (Milhões $)'].quantile(0.75)
limite_baixo = df['Prejuízo Financeiro (Milhões $)'].quantile(0.25)

def classifica_impacto(valor):
    if valor >= limite_alto:
        return 'Crítico (Alto Impacto)'
    elif valor <= limite_baixo:
        return 'Baixo'
    else:
        return 'Moderado'

df['Nível de Impacto'] = df['Prejuízo Financeiro (Milhões $)'].apply(classifica_impacto)
```

```python
# 2. Criando a coluna 'Custo por Usuário ($)'
# (Multiplicamos por 1 milhão para converter o prejuízo de milhões para dólares reais)
df['Custo por Usuário ($)'] = np.where(
    df['Usuários Afetados'] > 0,
    (df['Prejuízo Financeiro (Milhões $)' ] * 1_000_000) / df['Usuários Afetados'],
    0
)
```

```python
# 3. Criando a coluna 'Eficiência de Resolução' comparando com a média de horas
media_tempo = df['Tempo de Resolução (Horas)'].mean()
df['Eficiência de Resolução'] = df['Tempo de Resolução (Horas)'].apply(
    lambda x: 'Resposta Ágil (< Média)' if x <= media_tempo else 'Resposta Lenta (> Média)'
)
```
